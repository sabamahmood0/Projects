## Step 3: Handling Missing Values

### 3.1 Checking for Missing Values

```python

# Check for missing values
print(iris_df.isnull().sum())
```

### 3.2 Filling Numeric Missing Values

```python


# Fill missing values only in numeric columns with the mean of the column
iris_df.fillna(iris_df.select_dtypes(include='number').mean(), inplace=True)

# Confirm there are no missing values
print(iris_df.isnull().sum())
```

### 3.3 Filling Non-Numeric Missing Values



```python

# Get the mode of the 'Species' column
mode_value = iris_df['Species'].mode()[0]

# Assign the result back to the column
iris_df['Species'] = iris_df['Species'].fillna(mode_value)

# Confirm there are no missing values
print(iris_df.isnull().sum())


```

### 3.4 Dropping Missing Values

```python

# Drop rows with missing values
iris_df.dropna(inplace=True)
print(iris_df.head())
```

## Step 4: One-Hot Encoding

```python

# Perform one-hot encoding on the 'Species' column
iris_df_encoded = pd.get_dummies(iris_df, columns=['Species'])

# Display the encoded DataFrames

# First 5 entries of encoded dataframe
print("""
      
      First 5 entries of encoded dataframe
      
      """)
print(iris_df_encoded.head())



```

## Step 5: Normalization

### 5.1 Min-Max Normalization

```python

from sklearn.preprocessing import MinMaxScaler

# Initialize the MinMaxScaler
scaler = MinMaxScaler()

# Apply Min-Max scaling to 'SepalLengthCm'
iris_df[['SepalLengthCm']] = scaler.fit_transform(iris_df[['SepalLengthCm']])

# Display the normalized DataFrame
print(iris_df.head())
```

## Step 6: Adding and Removing Columns

### 6.1 Adding a New Column

```python

# Add a new column 'SepalArea' (SepalLengthCm * SepalWidthCm)
iris_df['SepalArea'] = iris_df['SepalLengthCm'] * iris_df['SepalWidthCm']
print(iris_df.head())
```

## Step 7: Merging Columns

### 7.1 Merging Two Columns

```python

# Merge 'SepalLengthCm' and 'SepalWidthCm' into 'SepalDimensions'
iris_df['SepalDimensions'] = iris_df['SepalLengthCm'].astype(str) + "x" + iris_df['SepalWidthCm'].astype(str)
print(iris_df[['SepalDimensions']].head())

print(""" 
      
      dataframe after adding new SepalDimensions Column 
      
      """)
print(iris_df.head())
```

## Step 8: Basic Descriptive Statistics
Pandas provides several methods to get basic descriptive statistics, such as:
- mean()
- median()
- std() (standard deviation)
- min(), max()
- count()

```python
# Calculate statistics for 'PetalLengthCm'
print("Mean Petal Length:", iris_df['PetalLengthCm'].mean())
print("Median Petal Length:", iris_df['PetalLengthCm'].median())
print("Standard Deviation of Petal Length:", iris_df['PetalLengthCm'].std())
print("Minimum Petal Length:", iris_df['PetalLengthCm'].min())
print("Maximum Petal Length:", iris_df['PetalLengthCm'].max())
print("Count of Petal Length:", iris_df['PetalLengthCm'].count())
```
