# Load the dataset

```python
import pandas as pd

# Load the dataset
file_path = "large_sample_dataset.csv"
df = pd.read_csv(file_path)

# Display the first few rows
df.head()
```

# Understand the dataset

```python
# Get basic information about the dataset
df.info()
```

```python
# Check for duplicate values
df.duplicated().sum()
```

```python
# Display summary statistics
df.describe(include='all')
```

# Handle missing values

```python
# Check missing values in each column
df.isnull().sum()
```

```python
# Fill missing values for numerical columns (e.g., Age, Income)
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Income'].fillna(df['Income'].median(), inplace=True)

# Fill missing values for categorical columns (e.g., Name, City, Subscription_Status)
df['City'].fillna(df['City'].mode()[0], inplace=True)
df['Subscription_Status'].fillna("Unknown", inplace=True)
```

# Handle duplicates

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)
```

# Standardize Data Formatting

```python
# Convert all string columns to lowercase for consistency
df['Name'] = df['Name'].str.title()
df['City'] = df['City'].str.title()
df['Subscription_Status'] = df['Subscription_Status'].str.capitalize()
```

# Convert Data Types

```python
# Convert Age and Income to integers
df['Age'] = df['Age'].astype(int)
df['Income'] = df['Income'].astype(int)
```

```python
df['Subscription_Status'].unique()
```

# Feature Encoding

```python
# Convert Gender into numerical values (Male = 0, Female = 1)

df['Gender'] = df['Gender'].map({'Male': 0, 'Female': 1})
# Convert Subscription_Status into numerical values
df['Subscription_Status'] = df['Subscription_Status'].map({'Active': 1, 'Inactive': 0, 'Unknown': -1})
```

```python
# Save the cleaned dataset to a new file
df.to_csv("cleaned_dataset.csv", index=False)
```
